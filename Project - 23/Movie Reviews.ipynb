### Preparing Data for ML ALGORITHMS

```python

from sklearn.preprocessing import OrdinalEncoder
encoder= OrdinalEncoder()
train_set[["sentiment"]]  =encoder.fit_transform(train_set[["sentiment"]])
test_set[["sentiment"]]=encoder.transform(test_set[["sentiment"]])
encoder.categories_
```

```python

train_set_final = train_set.drop("sentiment",axis=1)
train_set_label = train_set[["sentiment"]].to_numpy().ravel()
test_set_final = test_set.drop("sentiment",axis=1)
test_set_label = test_set[["sentiment"]].to_numpy().ravel()
```

```python

from sklearn.base import BaseEstimator,TransformerMixin
from scipy.sparse import csr_matrix
class Vectorizer(BaseEstimator,TransformerMixin):
    def __init__(self,size=1000):
        self.vocab_size=size
    def fit(self,X,y=None):
        GodCounter=Counter()
        PrevGodCounter=Counter()
        counters = self.get_counters(X)
        for i in counters:
            for word,count in i.most_common():
                PrevGodCounter[word]+=count
        for word,count in PrevGodCounter.most_common():
            GodCounter[word]=count
        self.vocab = {word:i for i,(word,count) in enumerate(GodCounter.most_common()[:self.vocab_size])}
        return self
    def transform(self,X,y=None):
        rows =[]
        data=[]
        cols=[]
        for row,i in enumerate(self.get_counters(X)):
            for word,count in i:
                rows.append(rows)
                cols.append(self.vocab.get(word,0))
                data.append(count)
        return csr_matrix((data,(rows,cols)),shape=(len(X),self.vocab_size+1))
    def get_counters(self,X):
        counters = []
        for j in range(len(X)):
            procces_i=re.sub(r'[^\w\s]','',re.sub(r'\d','NUMBER',X.iloc[j,0].lower()))
            counter = Counter([stemmer.stem(i) for i in procces_i.strip().split()])
        return counters
```

```python

vectorizer = Vectorizer()
train_set_transformed = vectorizer.fit_transform(train_set_final)
```
