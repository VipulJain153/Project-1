### Linear Algebra ALGORITHMS

**det(A)=∑(-1)^1+j x a1j x Mj1**

```python
import numpy as np
```

```python
train_set.to_numpy().shape
```

```python
coef=np.matmul(np.matmul(np.linalg.inv(np.matmul(train_set.to_numpy().T,train_set.to_numpy())),train_set.to_numpy().T),train_set_labels.to_numpy())
```

```python
coef
```

```python
# Ensure that coef is a column vector (shape: n_features, 1)
coef = coef.reshape(-1, 1)  # Make sure coef is a column vector

# Ensure test_set is in the correct format
test_set_numpy = test_set.to_numpy()  # Shape (n_samples, n_features)

# Make predictions
predictions = np.matmul(test_set_numpy, coef)  # (n_samples, n_features) x (n_features, 1)

# Predictions will have shape (n_samples, 1), you can flatten it to (n_samples,) if you want
predictions = predictions.flatten()

print("Predictions: ", predictions)
```

```python
import numpy as np

# Assuming you already have 'predictions' and 'test_labels' as numpy arrays

# Calculate the squared errors
squared_errors = (test_set_labels.to_numpy() - predictions) ** 2

# Compute the Mean Squared Error (MSE)
mse = squared_errors.mean()

print(f"Mean Squared Error: {mse}")
```
