```python
import sklearn.datasets as datasets
data = datasets.load_breast_cancer()
```

```python
print(data.DESCR)
```

```python
X = data.data
y = data.target
```

```python
from sklearn.svm import SVC
model=SVC()
model.fit(X,y)
```

```python
some_data =X[:3]
model.predict(some_data),y[:3]
```

# Cross - Validation

```python
from sklearn.model_selection import cross_val_predict
scores = cross_val_predict(model,X,y,cv=3,method="decision_function")
```

```python
from sklearn.metrics import precision_score
precision_score(model.predict(X),y)
```

```python
from sklearn.metrics import recall_score
recall_score(model.predict(X),y)
```

```python
from sklearn.metrics import confusion_matrix
confusion_matrix(model.predict(X),y)
```

## Regularization

```python
from sklearn.model_selection import RandomizedSearchCV
params = [{"tol":[.1,.2,.3,.4,.5,.6,.7,.8,.9,1]}]
SVModel = SVC()
random_search = RandomizedSearchCV(SVModel,params,cv=3,verbose=True,return_train_score=True)
```

```python
random_search.fit(X,y)
```

```python
random_search.best_estimator_
```

```python
precision_score(random_search.best_estimator_.predict(X),y)
```

```python
recall_score(random_search.best_estimator_.predict(X),y)
```

```python
from joblib import dump 
dump(random_search.best_estimator_,"model.joblib")
```
